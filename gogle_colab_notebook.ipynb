```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_excel("Cleaned_Sample_Dataset.xlsx")

# Basic information
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# Statistical Analysis
print("\n========== STATISTICAL SUMMARY ==========")
statistics = df.describe()
display(statistics)

print("\n========== MEDIAN ==========")
display(df.median(numeric_only=True))

print("\n========== STANDARD DEVIATION ==========")
display(df.std(numeric_only=True))

print("\n========== SKEWNESS ==========")
display(df.skew(numeric_only=True))

print("\n========== KURTOSIS ==========")
display(df.kurt(numeric_only=True))


# Product Analysis
product_analysis = df.groupby("product").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Average_Sales=("sales", "mean")
).sort_values("Total_Sales", ascending=False)

print("\n========== PRODUCT ANALYSIS ==========")
display(product_analysis)


# Category Analysis
category_analysis = df.groupby("category").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Average_Sales=("sales", "mean")
).sort_values("Total_Sales", ascending=False)

print("\n========== CATEGORY ANALYSIS ==========")
display(category_analysis)


# Region Analysis
region_analysis = df.groupby("region").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Average_Sales=("sales", "mean")
).sort_values("Total_Sales", ascending=False)

print("\n========== REGION ANALYSIS ==========")
display(region_analysis)


# Channel Analysis
channel_analysis = df.groupby("channel").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Average_Sales=("sales", "mean")
).sort_values("Total_Sales", ascending=False)

print("\n========== CHANNEL ANALYSIS ==========")
display(channel_analysis)


# Salesperson Analysis
salesperson_analysis = df.groupby("salesperson").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Average_Rating=("customer_rating", "mean")
).sort_values("Total_Sales", ascending=False)

print("\n========== SALESPERSON ANALYSIS ==========")
display(salesperson_analysis)


# Monthly Analysis
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["month"] = df["order_date"].dt.to_period("M").astype(str)

monthly_analysis = df.groupby("month").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity=("quantity", "sum"),
    Total_Orders=("order_id", "nunique")
).reset_index()

print("\n========== MONTHLY ANALYSIS ==========")
display(monthly_analysis)


# Correlation Analysis
numeric_columns = [
    "quantity",
    "unit_price",
    "discount",
    "sales",
    "profit",
    "customer_rating"
]

correlation = df[numeric_columns].corr()

print("\n========== CORRELATION ANALYSIS ==========")
display(correlation)


# Key Business Metrics
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_orders = df["order_id"].nunique()
total_quantity = df["quantity"].sum()
average_sales = df["sales"].mean()
average_profit = df["profit"].mean()
average_rating = df["customer_rating"].mean()

summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Quantity",
        "Average Sales",
        "Average Profit",
        "Average Customer Rating",
        "Best Product",
        "Best Category",
        "Best Region",
        "Best Channel",
        "Best Salesperson"
    ],
    "Value": [
        round(total_sales, 2),
        round(total_profit, 2),
        total_orders,
        total_quantity,
        round(average_sales, 2),
        round(average_profit, 2),
        round(average_rating, 2),
        product_analysis["Total_Sales"].idxmax(),
        category_analysis["Total_Sales"].idxmax(),
        region_analysis["Total_Sales"].idxmax(),
        channel_analysis["Total_Sales"].idxmax(),
        salesperson_analysis["Total_Sales"].idxmax()
    ]
})

print("\n========== KEY BUSINESS METRICS ==========")
display(summary)


# Top 10 Orders
top_10_orders = df.sort_values(
    "sales",
    ascending=False
).head(10)

print("\n========== TOP 10 ORDERS ==========")
display(top_10_orders)


# Save Analysis Results
output_file = "Python_Analysis_Results.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Data",
        index=False
    )

    statistics.to_excel(
        writer,
        sheet_name="Statistics"
    )

    product_analysis.to_excel(
        writer,
        sheet_name="Product_Analysis"
    )

    category_analysis.to_excel(
        writer,
        sheet_name="Category_Analysis"
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region_Analysis"
    )

    channel_analysis.to_excel(
        writer,
        sheet_name="Channel_Analysis"
    )

    salesperson_analysis.to_excel(
        writer,
        sheet_name="Salesperson_Analysis"
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly_Analysis",
        index=False
    )

    correlation.to_excel(
        writer,
        sheet_name="Correlation"
    )

    summary.to_excel(
        writer,
        sheet_name="Summary",
        index=False
    )

    top_10_orders.to_excel(
        writer,
        sheet_name="Top_10_Orders",
        index=False
    )

print("\n====================================")
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("====================================")
print("Output file: Python_Analysis_Results.xlsx")
print("Refresh the Colab Files panel to see the file.")
```
